##### PAN Column Inspection #####

```python
kyc_records_df["pan"].sample(10)
```

```python
kyc_records_df["pan"] = kyc_records_df["pan"].str.replace(" ", "").str.upper()
```

```python
print(kyc_records_df["pan"].isnull().sum())
print(kyc_records_df["pan"].duplicated().sum())
```

```python
import math
import regex as re

def clean_pan(value):
    if pd.isna(value):
        return np.nan

    value = re.sub( r"[^A-Z0-9]", "", value)

    if re.fullmatch(r"[A-Z]{5}[0-9]{4}[A-Z]", value):
        return value

    return np.nan

kyc_records_df["pan"] = kyc_records_df["pan"].apply(lambda x: clean_pan(x))
```

##### Adhaar Card Column Inspection #####

```python
kyc_records_df["aadhaar"].sample(10)
```

```python
kyc_records_df["aadhaar"] = kyc_records_df["aadhaar"].str.replace(" ", "").str.replace("-", "")
```

```python
print(kyc_records_df["aadhaar"].isnull().sum())
print(kyc_records_df["aadhaar"].duplicated().sum())
```

```python
def clean_aadhaar(val):
    if pd.isna(val):
        return np.nan

    if re.fullmatch(r"^\d{12}$", val):
        return val

    return np.nan

kyc_records_df["aadhaar"] = kyc_records_df["aadhaar"].apply(lambda x: clean_aadhaar(x))
```
